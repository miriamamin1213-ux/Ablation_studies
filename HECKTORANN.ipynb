```python
# ============================================================
# ANN — HECKTOR HPV CLASSIFIER | HANCOCK-ALIGNED | SEED 42 | NO SMOTE
# 500 Epochs | Stratified 3-Fold CV | Four Validation-Selected Checkpoints
# ============================================================
import os,json,random,copy,warnings,subprocess,sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import joblib
from torch.utils.data import TensorDataset,DataLoader
from google.colab import drive
from sklearn.model_selection import StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.metrics import (accuracy_score,balanced_accuracy_score,precision_score,
    recall_score,f1_score,roc_auc_score,average_precision_score,confusion_matrix,
    classification_report,roc_curve,precision_recall_curve)
warnings.filterwarnings("ignore",category=FutureWarning)
SEED=42
MAX_EPOCHS=500
BATCH_SIZE=32
N_SPLITS=3
# ------------------------------------------------------------
# 1. REPRODUCIBILITY, DEVICE AND DATA LOADING
# ------------------------------------------------------------
def seed_everything(seed=42):
    os.environ["PYTHONHASHSEED"]=str(seed)
    os.environ["CUBLAS_WORKSPACE_CONFIG"]=":4096:8"
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available(): torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic=True
    torch.backends.cudnn.benchmark=False
    torch.backends.cuda.matmul.allow_tf32=False
    torch.backends.cudnn.allow_tf32=False
    torch.use_deterministic_algorithms(True,warn_only=True)
seed_everything(SEED)
device=torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:",device)
drive.mount("/content/drive")
# ------------------------------------------------------------
# 2. LOAD HECKTOR DATA AND SELECT EIGHT PREDICTORS
# ------------------------------------------------------------
try:
    import gdown
except ImportError:
    subprocess.check_call([sys.executable,"-m","pip","install","-q","gdown"])
    import gdown
try:
    import shap
    from lime.lime_tabular import LimeTabularExplainer
except ImportError:
    subprocess.check_call([sys.executable,"-m","pip","install","-q","shap","lime"])
    import shap
    from lime.lime_tabular import LimeTabularExplainer
DATA_PATH="/content/HPV2025.xlsx"
if not os.path.exists(DATA_PATH):
    gdown.download(id="1DGDH2e6dEkpdwwtCvH6ds1CwJT2Hdmkv",output=DATA_PATH,quiet=False)
BASE="/content/drive/MyDrive"
OUT_DIR=os.path.join(BASE,"ANN_results_HECKTOR_3fold")
CHECKPOINT_DIR=os.path.join(OUT_DIR,"checkpoints")
os.makedirs(CHECKPOINT_DIR,exist_ok=True)
df=pd.read_excel(DATA_PATH)
if "HPV Status" not in df.columns: raise KeyError("Missing target column: HPV Status")
df=df.dropna(subset=["HPV Status"]).copy()
features=["Age","Gender","Tobacco Consumption","Alcohol Consumption",
          "Performance Status","T-stage","N-stage","M-stage"]
cat_cols=["Gender","Tobacco Consumption","Alcohol Consumption","Performance Status"]
num_cols=["Age","T-stage","N-stage","M-stage"]
required=features+["HPV Status"]
missing=[c for c in required if c not in df.columns]
if missing: raise KeyError(f"Missing required columns: {missing}")
df=df[required].copy()
stage_maps={"T-stage":{"T0":0,"T1":1,"T2":2,"T3":3,"T4":4},
            "N-stage":{"N0":0,"N1":1,"N2":2,"N3":3},"M-stage":{"M0":0,"M1":1}}
for col,mapping in stage_maps.items():
    df[col]=df[col].replace(mapping)
def normalize_label(v):
    if pd.isna(v): return np.nan
    s=str(v).strip().lower()
    if s in {"0","0.0","negative","neg","false","no"}: return 0
    if s in {"1","1.0","positive","pos","true","yes"}: return 1
    try:
        n=float(s)
        if n in (0.0,1.0): return int(n)
    except (ValueError,TypeError): pass
    return np.nan
df["HPV_Label"]=df["HPV Status"].map(normalize_label)
df=df.dropna(subset=["HPV_Label"]).copy()
df["HPV_Label"]=df["HPV_Label"].astype(int)
if set(df["HPV_Label"].unique())!={0,1}:
    raise ValueError("HPV Status must contain both binary classes coded as 0 and 1.")
df=df[features+["HPV_Label"]].reset_index(drop=True)
print("Rows:",len(df),"| Original predictors:",len(features))
print("Predictors:",features)
# ------------------------------------------------------------
# 3. STRATIFIED 3-FOLD CROSS-VALIDATION — NO TEST SET
# ------------------------------------------------------------
skf=StratifiedKFold(n_splits=N_SPLITS,shuffle=True,random_state=SEED)
thresholds=np.linspace(0.10,0.90,161)
checkpoint_specs={
    "best_val_loss":{"filename":"ann_hecktor_fold{fold}_best_val_loss.pt","direction":"min"},
    "best_val_roc_auc":{"filename":"ann_hecktor_fold{fold}_best_val_roc_auc.pt","direction":"max"},
    "best_val_pr_auc":{"filename":"ann_hecktor_fold{fold}_best_val_pr_auc.pt","direction":"max"},
    "best_val_balanced_accuracy":{
        "filename":"ann_hecktor_fold{fold}_best_val_balanced_accuracy.pt",
        "direction":"max"}}
def select_threshold(y_true,probabilities):
    scores=[balanced_accuracy_score(y_true,(probabilities>=t).astype(int)) for t in thresholds]
    best_score=max(scores)
    candidates=[t for t,s in zip(thresholds,scores) if np.isclose(s,best_score)]
    best_threshold=min(candidates,key=lambda t:abs(t-0.5))
    return float(best_threshold),float(best_score)
def predict_probabilities(model,X_tensor):
    model.eval()
    with torch.no_grad(): return torch.softmax(model(X_tensor),dim=1)[:,1].cpu().numpy()
def calculate_metrics(y_true,probabilities,threshold):
    pred=(probabilities>=threshold).astype(int)
    return {"Threshold":float(threshold),"Accuracy":accuracy_score(y_true,pred),
        "Balanced_Accuracy":balanced_accuracy_score(y_true,pred),
        "Precision":precision_score(y_true,pred,zero_division=0),
        "Recall":recall_score(y_true,pred,zero_division=0),
        "F1":f1_score(y_true,pred,zero_division=0),
        "ROC_AUC":roc_auc_score(y_true,probabilities) if len(np.unique(y_true))==2 else np.nan,
        "PR_AUC":average_precision_score(y_true,probabilities)}
# ------------------------------------------------------------
# 4. HANCOCK-ALIGNED ANN ARCHITECTURE
# ------------------------------------------------------------
class ANNClassifier(nn.Module):
    def __init__(self,input_dim):
        super().__init__()
        self.network=nn.Sequential(nn.Linear(input_dim,64),nn.ReLU(),nn.Dropout(0.30),
            nn.Linear(64,32),nn.ReLU(),nn.Dropout(0.20),nn.Linear(32,16),
            nn.ReLU(),nn.Linear(16,2))
    def forward(self,x): return self.network(x)
# ------------------------------------------------------------
# 5. TRAIN EACH FOLD — TRAIN-ONLY PREPROCESSING AND CLASS WEIGHTS
# ------------------------------------------------------------
fold_metrics=[]
fold_histories=[]
fold_predictions=[]
fold_models=[]
fold_shap_values=[]
fold_lime_values=[]
for fold,(train_idx,val_idx) in enumerate(skf.split(df,df["HPV_Label"]),start=1):
    print("\n"+"="*70,"\nFOLD",fold,"/",N_SPLITS,"\n"+"="*70)
    seed_everything(SEED+fold)
    train=df.iloc[train_idx].copy().reset_index(drop=True)
    val=df.iloc[val_idx].copy().reset_index(drop=True)
    print("Train:",len(train),"| Validation:",len(val))
    for name,part in [("Train",train),("Validation",val)]:
        print(f"{name} class counts:",part["HPV_Label"].value_counts().sort_index().to_dict())
    X_train_df=train[features].copy()
    X_val_df=val[features].copy()
    for frame in [X_train_df,X_val_df]:
        for col in num_cols: frame[col]=pd.to_numeric(frame[col],errors="coerce")
    for col in num_cols:
        median=X_train_df[col].median()
        if pd.isna(median): median=0.0
        for frame in [X_train_df,X_val_df]: frame[col]=frame[col].fillna(median)
    for col in cat_cols:
        for frame in [X_train_df,X_val_df]:
            frame[col]=frame[col].astype("object").replace(
                r"^\s*$",np.nan,regex=True).replace(["nan","None","NaN"],np.nan)
        modes=X_train_df[col].mode(dropna=True)
        fill_value=str(modes.iloc[0]) if not modes.empty else "Unknown"
        for frame in [X_train_df,X_val_df]:
            frame[col]=frame[col].fillna(fill_value).astype(str).replace(
                ["","nan","None","NaN"],"Unknown")
    try: encoder=OneHotEncoder(handle_unknown="ignore",sparse_output=False)
    except TypeError: encoder=OneHotEncoder(handle_unknown="ignore",sparse=False)
    preprocessor=ColumnTransformer([("num",StandardScaler(),num_cols),("cat",encoder,cat_cols)])
    X_train=preprocessor.fit_transform(X_train_df).astype(np.float32)
    X_val=preprocessor.transform(X_val_df).astype(np.float32)
    y_train=train["HPV_Label"].to_numpy(dtype=np.int64)
    y_val=val["HPV_Label"].to_numpy(dtype=np.int64)
    counts=np.bincount(y_train,minlength=2)
    if np.any(counts==0): raise ValueError("Both classes must be present in training data.")
    weights=len(y_train)/(2.0*counts)
    class_weights=torch.tensor(weights,dtype=torch.float32,device=device)
    print("Automatic class weights:",weights)
    print("SMOTE: False")
    train_dataset=TensorDataset(torch.tensor(X_train,dtype=torch.float32),
        torch.tensor(y_train,dtype=torch.long))
    train_loader=DataLoader(train_dataset,batch_size=BATCH_SIZE,shuffle=True,
        generator=torch.Generator().manual_seed(SEED+fold))
    X_val_t=torch.tensor(X_val,dtype=torch.float32,device=device)
    y_val_t=torch.tensor(y_val,dtype=torch.long,device=device)
    model=ANNClassifier(X_train.shape[1]).to(device)
    criterion=nn.CrossEntropyLoss(weight=class_weights,reduction="none")
    optimizer=torch.optim.Adam(model.parameters(),lr=0.001,weight_decay=1e-4)
    best_values={"best_val_loss":float("inf"),"best_val_roc_auc":-float("inf"),
        "best_val_pr_auc":-float("inf"),"best_val_balanced_accuracy":-float("inf")}
    best_epochs={}
    history=[]
    for epoch in range(1,MAX_EPOCHS+1):
        model.train()
        running_loss=0.0
        seen=0
        for xb,yb in train_loader:
            xb,yb=xb.to(device),yb.to(device)
            optimizer.zero_grad()
            per_example_loss=criterion(model(xb),yb)
            loss=per_example_loss.mean()
            loss.backward()
            optimizer.step()
            running_loss+=per_example_loss.detach().sum().item()
            seen+=len(yb)
        train_loss=running_loss/max(seen,1)
        model.eval()
        with torch.no_grad(): val_loss=criterion(model(X_val_t),y_val_t).mean().item()
        val_prob=predict_probabilities(model,X_val_t)
        val_threshold,val_bal_acc=select_threshold(y_val,val_prob)
        val_roc=roc_auc_score(y_val,val_prob) if len(np.unique(y_val))==2 else np.nan
        val_pr=average_precision_score(y_val,val_prob)
        history.append({"fold":fold,"epoch":epoch,"train_loss":float(train_loss),
            "val_loss":float(val_loss),"val_roc_auc":float(val_roc),
            "val_pr_auc":float(val_pr),"val_balanced_accuracy":float(val_bal_acc),
            "val_threshold":val_threshold})
        scores={"best_val_loss":val_loss,"best_val_roc_auc":val_roc,
            "best_val_pr_auc":val_pr,"best_val_balanced_accuracy":val_bal_acc}
        for key,spec in checkpoint_specs.items():
            score=float(scores[key])
            improved=score<best_values[key] if spec["direction"]=="min" else score>best_values[key]
            if improved:
                best_values[key]=score
                best_epochs[key]=epoch
                saved={"fold":fold,"epoch":epoch,
                    "model_state_dict":copy.deepcopy(model.state_dict()),
                    "validation_loss":float(val_loss),"validation_roc_auc":float(val_roc),
                    "validation_pr_auc":float(val_pr),
                    "validation_balanced_accuracy":float(val_bal_acc),
                    "validation_threshold":val_threshold}
                torch.save(saved,os.path.join(CHECKPOINT_DIR,
                    spec["filename"].format(fold=fold)))
        if epoch==1 or epoch%25==0:
            print(f"Epoch {epoch:03d}/{MAX_EPOCHS} | train loss={train_loss:.4f} "
                f"| val loss={val_loss:.4f} | val ROC-AUC={val_roc:.4f} "
                f"| val PR-AUC={val_pr:.4f} | val balanced accuracy={val_bal_acc:.4f}")
    if len(history)!=MAX_EPOCHS:
        raise RuntimeError(f"Expected {MAX_EPOCHS} epochs, but ran {len(history)}.")
    fold_histories.extend(history)
    history_df=pd.DataFrame(history)
    history_df.to_csv(os.path.join(OUT_DIR,f"training_history_fold{fold}.csv"),index=False)
    plt.figure(figsize=(9,5))
    plt.plot(history_df["epoch"],history_df["train_loss"],label="Train loss")
    plt.plot(history_df["epoch"],history_df["val_loss"],label="Validation loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(f"ANN — HECKTOR Fold {fold} Training History")
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR,f"training_history_fold{fold}.png"),dpi=150)
    plt.show()
    for checkpoint_name,spec in checkpoint_specs.items():
        checkpoint_path=os.path.join(CHECKPOINT_DIR,spec["filename"].format(fold=fold))
        try: checkpoint=torch.load(checkpoint_path,map_location=device,weights_only=False)
        except TypeError: checkpoint=torch.load(checkpoint_path,map_location=device)
        model.load_state_dict(checkpoint["model_state_dict"])
        val_prob=predict_probabilities(model,X_val_t)
        threshold,_=select_threshold(y_val,val_prob)
        metrics=calculate_metrics(y_val,val_prob,threshold)
        metrics.update({"Dataset":"HECKTOR","Model":"ANN","Fold":fold,
            "Checkpoint":checkpoint_name,"Checkpoint_Epoch":int(checkpoint["epoch"]),
            "Validation_Loss":float(checkpoint["validation_loss"]),
            "Validation_ROC_AUC":float(checkpoint["validation_roc_auc"]),
            "Validation_PR_AUC":float(checkpoint["validation_pr_auc"]),
            "Validation_Balanced_Accuracy":float(checkpoint["validation_balanced_accuracy"])})
        fold_metrics.append(metrics)
        fold_predictions.append(pd.DataFrame({"patient_index":val_idx,"y_true":y_val,
            "probability_positive":val_prob,"prediction":(val_prob>=threshold).astype(int),
            "fold":fold,"checkpoint":checkpoint_name,
            "checkpoint_epoch":int(checkpoint["epoch"]),"threshold":threshold}))
        print("\nCHECKPOINT:",checkpoint_name,"| Epoch:",checkpoint["epoch"],
            "\nFOLD VALIDATION METRICS\n",pd.Series(metrics).to_string(),
            "\nConfusion matrix:\n",confusion_matrix(y_val,(val_prob>=threshold).astype(int)),
            "\nClassification report:\n",classification_report(
                y_val,(val_prob>=threshold).astype(int),zero_division=0,digits=4))
    # Explain the balanced-accuracy-selected checkpoint in each fold.
    spec=checkpoint_specs["best_val_balanced_accuracy"]
    checkpoint=torch.load(os.path.join(CHECKPOINT_DIR,
        spec["filename"].format(fold=fold)),map_location=device,weights_only=False)
    model.load_state_dict(checkpoint["model_state_dict"])
    def predict_numpy(data):
        model.eval()
        tensor=torch.tensor(np.asarray(data,dtype=np.float32),device=device)
        with torch.no_grad(): return torch.softmax(model(tensor),dim=1).cpu().numpy()
    encoded_names=preprocessor.get_feature_names_out().tolist()
    encoded_to_original=[]
    for name in encoded_names:
        clean=name.split("__",1)[-1]
        original=next((f for f in features if clean==f or clean.startswith(f+"_")),None)
        if original is None:
            original=next((f for f in features if clean.startswith(f)),clean)
        encoded_to_original.append(original)
    background=shap.sample(pd.DataFrame(X_train,columns=encoded_names),
        min(40,len(X_train)),random_state=SEED+fold).to_numpy()
    explain_data=X_val[:min(100,len(X_val))]
    explainer=shap.KernelExplainer(lambda z:predict_numpy(z)[:,1],background)
    shap_raw=explainer.shap_values(explain_data,nsamples=100)
    if isinstance(shap_raw,list): shap_raw=shap_raw[0] if len(shap_raw)==1 else shap_raw[-1]
    shap_raw=np.asarray(shap_raw)
    if shap_raw.ndim==3: shap_raw=shap_raw[:,:,-1]
    grouped_shap=np.zeros((len(explain_data),len(features)))
    for j,feature in enumerate(features):
        cols=[i for i,name in enumerate(encoded_to_original) if name==feature]
        if cols: grouped_shap[:,j]=shap_raw[:,cols].sum(axis=1)
    fold_shap_values.append(grouped_shap)
    lime_explainer=LimeTabularExplainer(X_train,feature_names=encoded_names,
        class_names=["HPV negative","HPV positive"],mode="classification",
        discretize_continuous=False,random_state=SEED+fold)
    lime_abs=np.zeros(len(features))
    lime_n=min(50,len(X_val))
    for i in range(lime_n):
        exp=lime_explainer.explain_instance(X_val[i],predict_numpy,
            num_features=len(encoded_names),num_samples=1000)
        for encoded_idx,weight in exp.as_map().get(1,[]):
            lime_abs[features.index(encoded_to_original[encoded_idx])]+=abs(weight)
    fold_lime_values.append(lime_abs/max(lime_n,1))
    fold_models.append({"fold":fold,"preprocessor":preprocessor,
        "model_state_dict":copy.deepcopy(model.state_dict()),"features":features,
        "encoded_features":encoded_names})
    print("Fold",fold,"completed. Best checkpoint epochs:",best_epochs)
# ------------------------------------------------------------
# 6. THREE-FOLD METRICS, OOF PREDICTIONS AND CURVES
# ------------------------------------------------------------
metrics_df=pd.DataFrame(fold_metrics)
metrics_df.to_csv(os.path.join(OUT_DIR,"metrics_all_folds_checkpoints.csv"),index=False)
predictions_df=pd.concat(fold_predictions,ignore_index=True)
predictions_df.to_csv(os.path.join(OUT_DIR,"validation_predictions_all_folds.csv"),index=False)
metric_cols=["Accuracy","Balanced_Accuracy","Precision","Recall","F1","ROC_AUC","PR_AUC"]
summary=metrics_df.groupby("Checkpoint")[metric_cols].agg(["mean","std"])
summary.columns=[f"{a}_{b}" for a,b in summary.columns]
summary=summary.reset_index()
summary.to_csv(os.path.join(OUT_DIR,"metrics_mean_sd_3fold.csv"),index=False)
print("\n"+"="*90,"\nHECKTOR ANN — 3-FOLD VALIDATION SUMMARY (MEAN ± SD)\n"+"="*90)
print(summary.to_string(index=False))
for checkpoint_name in checkpoint_specs:
    part=predictions_df[predictions_df["checkpoint"]==checkpoint_name].sort_values("patient_index")
    y_true=part["y_true"].to_numpy()
    prob=part["probability_positive"].to_numpy()
    if len(np.unique(y_true))==2:
        fpr,tpr,_=roc_curve(y_true,prob)
        plt.figure(figsize=(8,6))
        plt.plot(fpr,tpr,label=f"{checkpoint_name} (AUC={roc_auc_score(y_true,prob):.3f})")
        plt.plot([0,1],[0,1],"--",label="Chance")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.title(f"ANN — HECKTOR 3-Fold OOF ROC: {checkpoint_name}")
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.savefig(os.path.join(OUT_DIR,f"roc_{checkpoint_name}.png"),dpi=150)
        plt.show()
        precision,recall,_=precision_recall_curve(y_true,prob)
        ap=average_precision_score(y_true,prob)
        plt.figure(figsize=(8,6))
        plt.plot(recall,precision,label=f"{checkpoint_name} (AP={ap:.3f})")
        plt.xlabel("Recall")
        plt.ylabel("Precision")
        plt.title(f"ANN — HECKTOR 3-Fold OOF Precision-Recall: {checkpoint_name}")
        plt.legend(fontsize=8)
        plt.tight_layout()
        plt.savefig(os.path.join(OUT_DIR,f"pr_{checkpoint_name}.png"),dpi=150)
        plt.show()
# ------------------------------------------------------------
# 7. MEAN SHAP AND LIME ACROSS THREE FOLDS
# ------------------------------------------------------------
mean_abs_shap=np.mean([np.mean(np.abs(x),axis=0) for x in fold_shap_values],axis=0)
mean_abs_lime=np.mean(fold_lime_values,axis=0)
shap_summary=pd.DataFrame({"Feature":features,"Mean_Absolute_SHAP":mean_abs_shap})
shap_summary=shap_summary.sort_values("Mean_Absolute_SHAP",ascending=False)
lime_summary=pd.DataFrame({"Feature":features,"Mean_Absolute_LIME":mean_abs_lime})
lime_summary=lime_summary.sort_values("Mean_Absolute_LIME",ascending=False)
shap_summary.to_csv(os.path.join(OUT_DIR,"mean_shap_importance_3fold.csv"),index=False)
lime_summary.to_csv(os.path.join(OUT_DIR,"mean_lime_importance_3fold.csv"),index=False)
print("\nMEAN ABSOLUTE SHAP IMPORTANCE — THREE FOLDS\n",shap_summary.to_string(index=False))
print("\nMEAN ABSOLUTE LIME IMPORTANCE — THREE FOLDS\n",lime_summary.to_string(index=False))
plt.figure(figsize=(9,6))
plt.barh(shap_summary["Feature"][::-1],shap_summary["Mean_Absolute_SHAP"][::-1])
plt.xlabel("Mean absolute SHAP value")
plt.title("ANN — Mean SHAP Importance Across 3 Folds")
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR,"mean_shap_importance_3fold.png"),dpi=150)
plt.show()
plt.figure(figsize=(9,6))
plt.barh(lime_summary["Feature"][::-1],lime_summary["Mean_Absolute_LIME"][::-1])
plt.xlabel("Mean absolute LIME weight")
plt.title("ANN — Mean LIME Importance Across 3 Folds")
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR,"mean_lime_importance_3fold.png"),dpi=150)
plt.show()
# SHAP beeswarm combines held-out validation explanations from all folds.
all_shap=np.vstack(fold_shap_values)
beeswarm_rows=[]
for _,val_idx in skf.split(df,df["HPV_Label"]):
    beeswarm_rows.append(df.iloc[val_idx][features].to_numpy()[:100])
beeswarm_data=pd.DataFrame(np.vstack(beeswarm_rows),columns=features)
for col in cat_cols:
    beeswarm_data[col]=pd.Categorical(beeswarm_data[col]).codes
for col in num_cols:
    beeswarm_data[col]=pd.to_numeric(beeswarm_data[col],errors="coerce")
shap.summary_plot(all_shap,features=beeswarm_data,feature_names=features,
    plot_type="dot",show=False,max_display=len(features))
plt.title("ANN — SHAP Beeswarm Across 3 Folds")
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR,"shap_beeswarm_3fold.png"),
    dpi=150,bbox_inches="tight")
plt.show()
# ------------------------------------------------------------
# 8. SAVE PREPROCESSORS, MODELS AND SETTINGS
# ------------------------------------------------------------
for item in fold_models:
    model_path=os.path.join(CHECKPOINT_DIR,
        f"ann_hecktor_fold{item['fold']}_final_explanation_model.pt")
    torch.save(item["model_state_dict"],model_path)
    prep_path=os.path.join(OUT_DIR,
        f"ann_hecktor_fold{item['fold']}_preprocessor.joblib")
    joblib.dump(item["preprocessor"],prep_path)
settings={
    "dataset":"HECKTOR","model":"ANN","seed":SEED,"n_splits":N_SPLITS,
    "features":features,"removed_features":["RFS","Treatment","Relapse"],
    "target_column":"HPV Status","smote":False,"early_stopping":False,
    "max_epochs":MAX_EPOCHS,"batch_size":BATCH_SIZE,
    "architecture":[len(features),64,32,16,2],"dropout":[0.30,0.20],
    "learning_rate":0.001,"weight_decay":1e-4,"optimizer":"Adam",
    "encoding":"OneHotEncoder","numeric_features":num_cols,
    "categorical_features":cat_cols,
    "validation_strategy":"Stratified 3-fold CV; no separate test set",
    "threshold_selection":"Maximize validation balanced accuracy; ties closest to 0.5",
    "checkpoint_epochs":"Saved separately for each fold"}
with open(os.path.join(OUT_DIR,"settings.json"),"w",encoding="utf-8") as f:
    json.dump(settings,f,indent=2)
print("\nSaved results to:",OUT_DIR)
print("Saved checkpoints to:",CHECKPOINT_DIR)
print("Output files:")
for root,dirs,files in os.walk(OUT_DIR):
    for filename in sorted(files): print(" -",os.path.relpath(os.path.join(root,filename),OUT_DIR))
```
