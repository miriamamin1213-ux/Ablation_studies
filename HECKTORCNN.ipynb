```python
# ============================================================
# CNN — HECKTOR HPV CLASSIFIER | HANCOCK-ALIGNED | SEED 42 | NO SMOTE
# 8 Predictors | Stratified 3-Fold CV | 500 Epochs | Four Checkpoints
# ============================================================
!pip -q install gdown shap lime
import os,json,random,copy,warnings,joblib,shutil
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import gdown
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset,DataLoader
from sklearn.model_selection import StratifiedKFold
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.metrics import (accuracy_score,balanced_accuracy_score,precision_score,
    recall_score,f1_score,roc_auc_score,average_precision_score,confusion_matrix,
    classification_report,roc_curve,precision_recall_curve)
import shap
from lime.lime_tabular import LimeTabularExplainer
warnings.filterwarnings("ignore",category=FutureWarning)
SEED=42
MAX_EPOCHS=500
BATCH_SIZE=32
N_SPLITS=3
# ------------------------------------------------------------
# 1. REPRODUCIBILITY AND DEVICE
# ------------------------------------------------------------
def seed_everything(seed=42):
    os.environ["PYTHONHASHSEED"]=str(seed)
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available(): torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic=True
    torch.backends.cudnn.benchmark=False
seed_everything(SEED)
device=torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:",device)
# ------------------------------------------------------------
# 2. DOWNLOAD AND LOAD HECKTOR DATA
# ------------------------------------------------------------
BASE="/content"
DATA_PATH="/content/HPV2025.xlsx"
OUT_DIR=os.path.join(BASE,"CNN_results_HECKTOR_3fold")
CHECKPOINT_DIR=os.path.join(OUT_DIR,"checkpoints")
os.makedirs(OUT_DIR,exist_ok=True)
os.makedirs(CHECKPOINT_DIR,exist_ok=True)
FILE_ID="1DGDH2e6dEkpdwwtCvH6ds1CwJT2Hdmkv"
if not os.path.isfile(DATA_PATH):
    downloaded=gdown.download(id=FILE_ID,output=DATA_PATH,quiet=False)
    if downloaded is None or not os.path.isfile(DATA_PATH):
        raise FileNotFoundError("Could not download HPV2025.xlsx.")
if not os.path.isfile(DATA_PATH) or os.path.getsize(DATA_PATH)==0:
    raise FileNotFoundError(f"Dataset is missing or empty: {DATA_PATH}")
df=pd.read_excel(DATA_PATH)
target_col="HPV Status"
if target_col not in df.columns: raise KeyError(f"Missing target: {target_col}")
df=df.dropna(subset=[target_col]).copy()
id_col=next((c for c in ["PatientID","patient_id"] if c in df.columns),None)
if id_col is None:
    patient_ids=pd.Series(np.arange(len(df)),index=df.index,name="patient_id")
else: patient_ids=df[id_col].copy()
df=df.drop(columns=["PatientID","patient_id","CenterID","Task 1","Task 2","Task 3"],
    errors="ignore")
features=["Age","Gender","Tobacco Consumption","Alcohol Consumption",
          "Performance Status","T-stage","N-stage","M-stage"]
numeric_features=["Age"]
categorical_features=["Gender","Tobacco Consumption","Alcohol Consumption",
                      "Performance Status","T-stage","N-stage","M-stage"]
missing=[c for c in features if c not in df.columns]
if missing: raise KeyError(f"Missing predictor columns: {missing}")
def normalize_label(value):
    if pd.isna(value): return np.nan
    value_str=str(value).strip().lower()
    if value_str in {"negative","neg","0","0.0","false","no"}: return 0
    if value_str in {"positive","pos","1","1.0","true","yes"}: return 1
    try:
        number=float(value_str)
        if number in (0.0,1.0): return int(number)
    except (ValueError,TypeError): pass
    return np.nan
df["HPV_Label"]=df[target_col].map(normalize_label)
df=df.dropna(subset=["HPV_Label"]).copy()
df["HPV_Label"]=df["HPV_Label"].astype(int)
if df["HPV_Label"].nunique()!=2:
    raise ValueError("HPV Status must contain both classes, 0 and 1.")
df=df[features+["HPV_Label"]].copy()
df["patient_id"]=patient_ids.reindex(df.index).values
df=df.reset_index(drop=True)
print("Rows available:",len(df))
print("Predictors:",features)
print("Target class counts:\n",df["HPV_Label"].value_counts().sort_index())
print("SMOTE: False")
# ------------------------------------------------------------
# 3. STRATIFIED 3-FOLD CROSS-VALIDATION — NO TEST SET
# ------------------------------------------------------------
skf=StratifiedKFold(n_splits=N_SPLITS,shuffle=True,random_state=SEED)
thresholds=np.linspace(0.10,0.90,161)
checkpoint_specs={
    "best_val_loss":{"filename":"cnn_fold{fold}_best_val_loss.pt","direction":"min"},
    "best_val_roc_auc":{"filename":"cnn_fold{fold}_best_val_roc_auc.pt","direction":"max"},
    "best_val_pr_auc":{"filename":"cnn_fold{fold}_best_val_pr_auc.pt","direction":"max"},
    "best_val_balanced_accuracy":{
        "filename":"cnn_fold{fold}_best_val_balanced_accuracy.pt","direction":"max"}}
def select_threshold(y_true,probabilities):
    scores=[balanced_accuracy_score(y_true,(probabilities>=t).astype(int)) for t in thresholds]
    best=max(scores)
    chosen=min((t for t,s in zip(thresholds,scores) if np.isclose(s,best)),
        key=lambda t:abs(t-0.5))
    return float(chosen),float(best)
def predict_probabilities(model,X_tensor):
    model.eval()
    with torch.no_grad():
        return torch.softmax(model(X_tensor),dim=1)[:,1].cpu().numpy()
def calculate_metrics(y_true,probabilities,threshold):
    pred=(probabilities>=threshold).astype(int)
    return {"Threshold":float(threshold),"Accuracy":accuracy_score(y_true,pred),
        "Balanced_Accuracy":balanced_accuracy_score(y_true,pred),
        "Precision":precision_score(y_true,pred,zero_division=0),
        "Recall":recall_score(y_true,pred,zero_division=0),
        "F1":f1_score(y_true,pred,zero_division=0),
        "ROC_AUC":roc_auc_score(y_true,probabilities),
        "PR_AUC":average_precision_score(y_true,probabilities)}
# ------------------------------------------------------------
# 4. HANCOCK-ALIGNED CNN ARCHITECTURE
# ------------------------------------------------------------
class CNNClassifier(nn.Module):
    def __init__(self,input_dim):
        super().__init__()
        self.conv=nn.Conv1d(1,16,kernel_size=3,padding=1)
        self.relu=nn.ReLU()
        self.dropout=nn.Dropout(0.25)
        self.fc1=nn.Linear(input_dim*16,32)
        self.fc2=nn.Linear(32,16)
        self.fc3=nn.Linear(16,2)
    def forward(self,x):
        x=x.unsqueeze(1)
        x=self.conv(x)
        x=self.relu(x)
        x=self.dropout(x)
        x=x.flatten(start_dim=1)
        x=self.fc1(x)
        x=self.relu(x)
        x=self.fc2(x)
        x=self.relu(x)
        return self.fc3(x)
# ------------------------------------------------------------
# 5. TRAIN EACH FOLD — TRAIN-ONLY PREPROCESSING AND CLASS WEIGHTS
# ------------------------------------------------------------
fold_metrics=[]
fold_histories=[]
fold_predictions={name:[] for name in checkpoint_specs}
fold_shap_values=[]
fold_shap_features=[]
fold_lime_values=[]
fold_models=[]
for fold,(train_idx,val_idx) in enumerate(skf.split(df,df["HPV_Label"]),start=1):
    print("\n"+"="*70,"\nFOLD",fold,"/",N_SPLITS,"\n"+"="*70)
    seed_everything(SEED+fold)
    train=df.iloc[train_idx].copy().reset_index(drop=True)
    val=df.iloc[val_idx].copy().reset_index(drop=True)
    for name,part in [("Train",train),("Validation",val)]:
        print(f"{name}: {len(part)} rows | classes: "
              f"{part['HPV_Label'].value_counts().sort_index().to_dict()}")
    X_train_df=train[features].copy()
    X_val_df=val[features].copy()
    for frame in [X_train_df,X_val_df]:
        for col in numeric_features:
            frame[col]=pd.to_numeric(frame[col],errors="coerce")
        for col in categorical_features:
            frame[col]=frame[col].astype("object").replace(
                r"^\s*$",np.nan,regex=True).replace(["nan","None","NaN"],np.nan)
    for col in numeric_features:
        median=X_train_df[col].median()
        if pd.isna(median): median=0.0
        for frame in [X_train_df,X_val_df]: frame[col]=frame[col].fillna(median)
    for col in categorical_features:
        modes=X_train_df[col].mode(dropna=True)
        fill=str(modes.iloc[0]) if not modes.empty else "Unknown"
        for frame in [X_train_df,X_val_df]:
            frame[col]=frame[col].fillna(fill).astype(str).replace(
                ["","nan","None","NaN"],"Unknown")
    try: encoder=OneHotEncoder(handle_unknown="ignore",sparse_output=False)
    except TypeError: encoder=OneHotEncoder(handle_unknown="ignore",sparse=False)
    preprocessor=ColumnTransformer([
        ("numeric",StandardScaler(),numeric_features),
        ("categorical",encoder,categorical_features)])
    X_train=preprocessor.fit_transform(X_train_df).astype(np.float32)
    X_val=preprocessor.transform(X_val_df).astype(np.float32)
    y_train=train["HPV_Label"].to_numpy(dtype=np.int64)
    y_val=val["HPV_Label"].to_numpy(dtype=np.int64)
    counts=np.bincount(y_train,minlength=2)
    if np.any(counts==0): raise ValueError("Both classes must be present in training.")
    weights=len(y_train)/(2.0*counts)
    class_weights=torch.tensor(weights,dtype=torch.float32,device=device)
    print("Automatic class weights:",weights)
    print("SMOTE: False")
    train_dataset=TensorDataset(torch.tensor(X_train,dtype=torch.float32),
        torch.tensor(y_train,dtype=torch.long))
    train_loader=DataLoader(train_dataset,batch_size=BATCH_SIZE,shuffle=True,
        generator=torch.Generator().manual_seed(SEED+fold))
    X_val_t=torch.tensor(X_val,dtype=torch.float32,device=device)
    y_val_t=torch.tensor(y_val,dtype=torch.long,device=device)
    model=CNNClassifier(X_train.shape[1]).to(device)
    criterion=nn.CrossEntropyLoss(weight=class_weights,reduction="none")
    optimizer=torch.optim.Adam(model.parameters(),lr=0.0001,weight_decay=1e-4)
    best_values={"best_val_loss":float("inf"),"best_val_roc_auc":-float("inf"),
        "best_val_pr_auc":-float("inf"),"best_val_balanced_accuracy":-float("inf")}
    best_epochs={}
    history=[]
    for epoch in range(1,MAX_EPOCHS+1):
        model.train()
        running_loss=0.0
        seen=0
        for xb,yb in train_loader:
            xb,yb=xb.to(device),yb.to(device)
            optimizer.zero_grad()
            per_example_loss=criterion(model(xb),yb)
            loss=per_example_loss.mean()
            loss.backward()
            optimizer.step()
            running_loss+=per_example_loss.detach().sum().item()
            seen+=len(yb)
        train_loss=running_loss/max(seen,1)
        model.eval()
        with torch.no_grad(): val_loss=criterion(model(X_val_t),y_val_t).mean().item()
        val_prob=predict_probabilities(model,X_val_t)
        val_threshold,val_bal=select_threshold(y_val,val_prob)
        val_roc=roc_auc_score(y_val,val_prob)
        val_pr=average_precision_score(y_val,val_prob)
        history.append({"fold":fold,"epoch":epoch,"train_loss":float(train_loss),
            "val_loss":float(val_loss),"val_roc_auc":float(val_roc),
            "val_pr_auc":float(val_pr),"val_balanced_accuracy":float(val_bal),
            "val_threshold":val_threshold})
        scores={"best_val_loss":val_loss,"best_val_roc_auc":val_roc,
            "best_val_pr_auc":val_pr,"best_val_balanced_accuracy":val_bal}
        for name,spec in checkpoint_specs.items():
            score=float(scores[name])
            improved=(score<best_values[name] if spec["direction"]=="min"
                      else score>best_values[name])
            if improved:
                best_values[name]=score
                best_epochs[name]=epoch
                torch.save({"model_state_dict":copy.deepcopy(model.state_dict()),
                    "epoch":epoch,"validation_loss":float(val_loss),
                    "validation_roc_auc":float(val_roc),"validation_pr_auc":float(val_pr),
                    "validation_balanced_accuracy":float(val_bal),
                    "validation_threshold":val_threshold},
                    os.path.join(CHECKPOINT_DIR,spec["filename"].format(fold=fold)))
        if epoch==1 or epoch%25==0:
            print(f"Epoch {epoch:03d}/{MAX_EPOCHS} | train loss={train_loss:.4f} "
                f"| val loss={val_loss:.4f} | ROC-AUC={val_roc:.4f} "
                f"| PR-AUC={val_pr:.4f} | balanced accuracy={val_bal:.4f}")
    if len(history)!=MAX_EPOCHS:
        raise RuntimeError(f"Expected {MAX_EPOCHS} epochs, ran {len(history)}.")
    fold_histories.extend(history)
    history_df=pd.DataFrame(history)
    history_df.to_csv(os.path.join(OUT_DIR,f"training_history_fold{fold}.csv"),index=False)
    plt.figure(figsize=(9,5))
    plt.plot(history_df["epoch"],history_df["train_loss"],label="Train loss")
    plt.plot(history_df["epoch"],history_df["val_loss"],label="Validation loss")
    plt.xlabel("Epoch")
    plt.ylabel("Loss")
    plt.title(f"CNN — HECKTOR Fold {fold} Training History")
    plt.legend()
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR,f"training_history_fold{fold}.png"),dpi=150)
    plt.show()
    for name,spec in checkpoint_specs.items():
        path=os.path.join(CHECKPOINT_DIR,spec["filename"].format(fold=fold))
        try: saved=torch.load(path,map_location=device,weights_only=False)
        except TypeError: saved=torch.load(path,map_location=device)
        model.load_state_dict(saved["model_state_dict"])
        prob=predict_probabilities(model,X_val_t)
        threshold,_=select_threshold(y_val,prob)
        metrics=calculate_metrics(y_val,prob,threshold)
        metrics.update({"Dataset":"HECKTOR","Model":"CNN","Fold":fold,
            "Checkpoint":name,"Checkpoint_Epoch":int(saved["epoch"]),
            "Validation_Loss":float(saved["validation_loss"]),
            "Validation_ROC_AUC":float(saved["validation_roc_auc"]),
            "Validation_PR_AUC":float(saved["validation_pr_auc"]),
            "Validation_Balanced_Accuracy":float(saved["validation_balanced_accuracy"])})
        fold_metrics.append(metrics)
        pred=(prob>=threshold).astype(int)
        fold_predictions[name].append(pd.DataFrame({"patient_id":val["patient_id"],
            "patient_index":val_idx,"y_true":y_val,"probability_positive":prob,
            "prediction":pred,"fold":fold,"checkpoint":name,
            "checkpoint_epoch":int(saved["epoch"]),"threshold":threshold}))
        print("\nCHECKPOINT:",name,"| Epoch:",saved["epoch"],
            "\nFOLD VALIDATION METRICS\n",pd.Series(metrics).to_string(),
            "\nConfusion matrix:\n",confusion_matrix(y_val,pred),
            "\nClassification report:\n",classification_report(
                y_val,pred,zero_division=0,digits=4))
    # SHAP and LIME explain the balanced-accuracy-selected fold checkpoint.
    spec=checkpoint_specs["best_val_balanced_accuracy"]
    saved=torch.load(os.path.join(CHECKPOINT_DIR,
        spec["filename"].format(fold=fold)),map_location=device,weights_only=False)
    model.load_state_dict(saved["model_state_dict"])
    def predict_numpy(data):
        tensor=torch.tensor(np.asarray(data,dtype=np.float32),device=device)
        model.eval()
        with torch.no_grad(): return torch.softmax(model(tensor),dim=1).cpu().numpy()
    encoded_names=preprocessor.get_feature_names_out().tolist()
    encoded_to_original=[]
    for name in encoded_names:
        clean=name.split("__",1)[-1]
        match=next((f for f in features if clean==f or clean.startswith(f+"_")),None)
        if match is None: match=next((f for f in features if clean.startswith(f)),clean)
        encoded_to_original.append(match)
    background=shap.sample(pd.DataFrame(X_train,columns=encoded_names),
        min(40,len(X_train)),random_state=SEED+fold).to_numpy()
    explain_data=X_val[:min(100,len(X_val))]
    explainer=shap.KernelExplainer(lambda z:predict_numpy(z)[:,1],background)
    raw_shap=explainer.shap_values(explain_data,nsamples=100)
    if isinstance(raw_shap,list): raw_shap=raw_shap[-1]
    raw_shap=np.asarray(raw_shap)
    if raw_shap.ndim==3: raw_shap=raw_shap[:,:,-1]
    grouped=np.zeros((len(explain_data),len(features)))
    for j,feature in enumerate(features):
        cols=[i for i,value in enumerate(encoded_to_original) if value==feature]
        if cols: grouped[:,j]=raw_shap[:,cols].sum(axis=1)
    fold_shap_values.append(grouped)
    fold_shap_features.append(val[features].iloc[:len(explain_data)].copy())
    lime=LimeTabularExplainer(X_train,feature_names=encoded_names,
        class_names=["HPV negative","HPV positive"],mode="classification",
        discretize_continuous=False,random_state=SEED+fold)
    lime_abs=np.zeros(len(features))
    lime_n=min(50,len(X_val))
    for i in range(lime_n):
        exp=lime.explain_instance(X_val[i],predict_numpy,
            num_features=len(encoded_names),num_samples=1000)
        for idx,weight in exp.as_map().get(1,[]):
            lime_abs[features.index(encoded_to_original[idx])]+=abs(weight)
    fold_lime_values.append(lime_abs/max(lime_n,1))
    fold_models.append({"fold":fold,"model_state_dict":copy.deepcopy(model.state_dict()),
        "preprocessor":preprocessor,"features":features,"encoded_features":encoded_names})
    print("Fold",fold,"completed. Best epochs:",best_epochs)
# ------------------------------------------------------------
# 6. THREE-FOLD METRICS, OOF PREDICTIONS AND CURVES
# ------------------------------------------------------------
metrics_df=pd.DataFrame(fold_metrics)
metrics_df.to_csv(os.path.join(OUT_DIR,"metrics_all_folds_checkpoints.csv"),index=False)
summary_cols=["Accuracy","Balanced_Accuracy","Precision","Recall","F1","ROC_AUC","PR_AUC"]
summary=metrics_df.groupby("Checkpoint")[summary_cols].agg(["mean","std"])
summary.columns=[f"{a}_{b}" for a,b in summary.columns]
summary=summary.reset_index()
summary.to_csv(os.path.join(OUT_DIR,"metrics_mean_sd_3fold.csv"),index=False)
print("\nCNN — HECKTOR 3-FOLD VALIDATION SUMMARY (MEAN ± SD)\n",
    summary.to_string(index=False))
for name in checkpoint_specs:
    part=pd.concat(fold_predictions[name],ignore_index=True).sort_values("patient_index")
    part.to_csv(os.path.join(OUT_DIR,f"validation_predictions_{name}.csv"),index=False)
    y_true=part["y_true"].to_numpy()
    prob=part["probability_positive"].to_numpy()
    fpr,tpr,_=roc_curve(y_true,prob)
    plt.figure(figsize=(8,6))
    plt.plot(fpr,tpr,label=f"{name} (AUC={roc_auc_score(y_true,prob):.3f})")
    plt.plot([0,1],[0,1],"--",label="Chance")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"CNN — HECKTOR 3-Fold OOF ROC: {name}")
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR,f"roc_{name}.png"),dpi=150)
    plt.show()
    precision,recall,_=precision_recall_curve(y_true,prob)
    plt.figure(figsize=(8,6))
    plt.plot(recall,precision,label=f"{name} (AP={average_precision_score(y_true,prob):.3f})")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title(f"CNN — HECKTOR 3-Fold OOF Precision-Recall: {name}")
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR,f"pr_{name}.png"),dpi=150)
    plt.show()
# ------------------------------------------------------------
# 7. MEAN SHAP, LIME AND SHAP BEESWARM ACROSS THREE FOLDS
# ------------------------------------------------------------
mean_shap=np.mean([np.mean(np.abs(x),axis=0) for x in fold_shap_values],axis=0)
mean_lime=np.mean(fold_lime_values,axis=0)
shap_summary=pd.DataFrame({"Feature":features,"Mean_Absolute_SHAP":mean_shap})
shap_summary=shap_summary.sort_values("Mean_Absolute_SHAP",ascending=False)
lime_summary=pd.DataFrame({"Feature":features,"Mean_Absolute_LIME":mean_lime})
lime_summary=lime_summary.sort_values("Mean_Absolute_LIME",ascending=False)
shap_summary.to_csv(os.path.join(OUT_DIR,"mean_shap_importance_3fold.csv"),index=False)
lime_summary.to_csv(os.path.join(OUT_DIR,"mean_lime_importance_3fold.csv"),index=False)
print("\nMEAN SHAP IMPORTANCE\n",shap_summary.to_string(index=False))
print("\nMEAN LIME IMPORTANCE\n",lime_summary.to_string(index=False))
for data,label,title,filename in [
    (shap_summary,"Mean absolute SHAP value","Mean SHAP Importance",
     "mean_shap_importance_3fold.png"),
    (lime_summary,"Mean absolute LIME weight","Mean LIME Importance",
     "mean_lime_importance_3fold.png")]:
    plt.figure(figsize=(9,6))
    plt.barh(data["Feature"][::-1],data.iloc[::-1,1])
    plt.xlabel(label)
    plt.title(f"CNN — {title} Across 3 Folds")
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR,filename),dpi=150)
    plt.show()
all_shap=np.vstack(fold_shap_values)
all_features=pd.concat(fold_shap_features,ignore_index=True)
beeswarm_data=all_features.copy()
for col in categorical_features:
    beeswarm_data[col]=pd.Categorical(beeswarm_data[col]).codes
beeswarm_data["Age"]=pd.to_numeric(beeswarm_data["Age"],errors="coerce")
shap.summary_plot(all_shap,features=beeswarm_data,feature_names=features,
    plot_type="dot",show=False,max_display=len(features))
plt.title("CNN — SHAP Beeswarm Across 3 Folds")
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR,"shap_beeswarm_3fold.png"),dpi=150,bbox_inches="tight")
plt.show()
# ------------------------------------------------------------
# 8. SAVE FOLD MODELS, PREPROCESSORS, SETTINGS AND RESULTS ZIP
# ------------------------------------------------------------
for item in fold_models:
    torch.save(item["model_state_dict"],os.path.join(CHECKPOINT_DIR,
        f"cnn_fold{item['fold']}_explanation_model.pt"))
    joblib.dump(item["preprocessor"],os.path.join(OUT_DIR,
        f"cnn_fold{item['fold']}_preprocessor.joblib"))
settings={"model":"CNN","dataset":"HECKTOR","seed":SEED,"n_splits":N_SPLITS,
    "features":features,"numeric_features":numeric_features,
    "categorical_features":categorical_features,
    "removed_features":["Relapse","RFS","Treatment"],"target_column":target_col,
    "target_mapping":{"Negative":0,"Positive":1},"smote":False,
    "early_stopping":False,"max_epochs":MAX_EPOCHS,"epochs_per_fold":MAX_EPOCHS,
    "batch_size":BATCH_SIZE,
    "architecture":"Conv1d(1,16,kernel_size=3,padding=1)-Dropout(0.25)-FC32-FC16-FC2",
    "optimizer":"Adam","learning_rate":0.0001,"weight_decay":0.0001,
    "validation_strategy":"Stratified 3-fold CV; no separate test set",
    "threshold_selection":"Maximize validation balanced accuracy; ties closest to 0.5",
    "checkpoint_epochs":"Saved separately for each fold"}
with open(os.path.join(OUT_DIR,"settings.json"),"w",encoding="utf-8") as f:
    json.dump(settings,f,indent=2)
RESULTS_ZIP=shutil.make_archive(OUT_DIR,"zip",root_dir=OUT_DIR)
print("\nSaved results to:",OUT_DIR)
print("Results archive:",RESULTS_ZIP)
print("Download the ZIP before resetting the Colab runtime.")
```
